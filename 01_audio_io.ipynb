# 01 · Audio I/O: what the Realtime API actually receives

A "live" voice model never sees a WAV file. It sees a **stream of small base64 chunks of raw PCM16 audio**.
This notebook follows one recording from the microphone all the way to the exact JSON event that the Realtime API would receive.

`mic → int16 samples → PCM16 bytes → chunks → base64 → {"type": "input_audio_buffer.append"}`

No API calls here. Helpers live in `live/audio.py`.

```python
import json
import numpy as np
import sounddevice as sd
import matplotlib.pyplot as plt
from IPython.display import Audio
from live import audio
```

## 1. Which devices will be used?
`sd.default.device` = (input index, output index). **Use headphones** in later notebooks, so the model doesn't hear itself.

```python
devices = sd.query_devices()
default_in, default_out = sd.default.device
devices[default_in]['name'], devices[default_out]['name']
```

## 2. The audio format the API expects
OpenAI Realtime uses 24 kHz, mono, 16-bit signed integers (PCM16), for both input and output.

```python
audio.RATE, audio.CHANNELS, audio.DTYPE
```

## 3. Record 3 seconds (say something!)
The result is just a numpy array of integers. Each one is the mic membrane position at one instant, from -32768 to 32767.

```python
rec = audio.record(3)
rec.shape, rec.dtype, rec.min(), rec.max()
```

Number of samples = seconds × rate:

```python
len(rec), 3 * audio.RATE
```

**Loudness (RMS)**: silence is about 0.00x, speech about 0.02–0.2. Server-side VAD uses a signal like this to decide that you are talking.

```python
audio.rms(rec)
```

```python
t = np.arange(len(rec)) / audio.RATE
plt.figure(figsize=(10, 2.5)); plt.plot(t, rec, lw=0.4)
plt.xlabel('seconds'); plt.ylabel('int16 sample'); plt.title('Your 3 s recording'); plt.show()
```

```python
Audio(rec, rate=audio.RATE)  # in-notebook player
```

## 4. int16 ↔ float
DSP code (and our RMS) works on floats in [-1, 1]. The wire format is int16. Converting back and forth loses at most 1 unit, because of rounding.

```python
f = audio.pcm16_to_float(rec)
f.dtype, round(float(f.min()), 4), round(float(f.max()), 4)
```

```python
back = audio.float_to_pcm16(f)
int(np.abs(back.astype(int) - rec).max())  # max round-trip error, in int16 units
```

## 5. Samples → PCM16 bytes
Each sample takes 2 bytes, little-endian. That is all "PCM16" means: no header, no compression.

```python
pcm = audio.to_bytes(rec)
len(pcm), len(rec) * 2, pcm[:8]
```

## 6. Bytes → chunks
We stream audio as it is captured instead of sending it all at once. Each chunk is a short slice of time. 100 ms is a good balance: smaller chunks mean lower latency but more messages.

```python
CHUNK_MS = 100
chunk_bytes = audio.RATE * CHUNK_MS // 1000 * 2   # samples per chunk × 2 bytes
chunks = [pcm[i:i + chunk_bytes] for i in range(0, len(pcm), chunk_bytes)]
chunk_bytes, len(chunks)
```

## 7. Chunk → base64
JSON can't carry raw bytes, so each chunk is base64-encoded. That makes it about 33% bigger.

```python
b64 = audio.to_b64(chunks[0])
len(chunks[0]), len(b64), b64[:60]
```

## 8. The exact event sent over the WebSocket
Notebook 03 sends one of these every 100 ms while you talk:

```python
event = {"type": "input_audio_buffer.append", "audio": b64}
print(json.dumps(event)[:120] + ' ...')
len(json.dumps(event))
```

## 9. The reverse path (what we'll do with the model's audio)
The server sends `response.output_audio.delta` events with the same base64 PCM16. Decoding gives back the identical samples:

```python
decoded = audio.from_bytes(audio.from_b64(b64))
assert np.array_equal(decoded, rec[:len(decoded)])
decoded.shape, decoded.dtype
```

```python
audio.play(rec)  # through the speakers via sounddevice: this is how model audio will be played
```

## Takeaways
- The API's audio is just `int16` numbers at 24 kHz. There is nothing more to it.
- "Streaming" = cut the byte stream into ~100 ms chunks and send each one as a base64 JSON event.
- The output path is the mirror image: base64 → bytes → int16 → speaker.
